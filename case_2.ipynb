```python
# cleaning kolom description aja
def clean_descrip(text):
    # lowercase
    text = text.lower()
    # url
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    # angka dan tanda baca
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    # digit angka
    text = re.sub(r'\d+', '', text)
    # hashtag dan mention
    text = re.sub(r'@\w+|#\w+', '', text)
    # spasi berlebih
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text
    

df['cleaning'] = df['full_text'].astype(str).apply(clean_descrip)

display(df[['full_text', 'cleaning']].head())
```

```python
def tokenize_text(text):
    if isinstance(text, str):
        return text.split()
    else:
        return []

df['tokens'] = df['cleaning'].apply(tokenize_text)

display(df[['full_text', 'cleaning', 'tokens']].head())
```

```python
stopwords = [
    'dan','di','yang','ke','dari','untuk',
    'ini','itu','atau','karena','ada',
    'the','and','of','to','is','in',
    'a','i','you','my','me'
]

def remove_stopwords(tokens):

    hasil = []

    for word in tokens:
        if word not in stopwords:
            hasil.append(word)

    return hasil


df['stopwords_removed'] = df['tokens'].apply(remove_stopwords)

display(df[['tokens', 'stopwords_removed']].head())
```

```python
features = vectorizer.get_feature_names_out()

# Buat DataFrame TF-IDF, baris = dokumen, kolom = fitur
tfidf_df = pd.DataFrame(X_tfidf.toarray(), columns=features)

# Ubah nama kolom jadi D1, D2, dst untuk dokumen (opsional)
tfidf_df.index = [f'D{i+1}' for i in range(tfidf_df.shape[0])]

# Transpose supaya term jadi baris dan dokumen jadi kolom
tfidf_table = tfidf_df.T
tfidf_table.columns.name = "Document"
tfidf_table.index.name = "Term"

# Tampilkan
# Hitung rata-rata TF-IDF tiap term di semua dokumen
term_means = tfidf_table.mean(axis=1)
# Pilih top 20 term
top_terms = term_means.sort_values(ascending=False).head(20).index
# Tampilkan hanya term tersebut
display(tfidf_table.loc[top_terms].round(4))
```

```python
from sklearn.decomposition import LatentDirichletAllocation

# model LDA
lda = LatentDirichletAllocation(
    n_components=5,
    random_state=42
)

# training LDA pakai TF-IDF
lda.fit(X_tfidf)

print("LDA berhasil dibuat")
```

```python
# ambil nama fitur/kata
fitur = vectorizer.get_feature_names_out()

# tampilkan topik
for i, topik in enumerate(lda.components_):

    print(f"\nTopik {i+1}")

    kata_topik = [
        fitur[index]
        for index in topik.argsort()[-10:]
    ]

    print(kata_topik)
```
